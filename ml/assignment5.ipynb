```python
pip install Minisom 
import numpy as np 
import matplotlib.pyplot as plt 
from sklearn.datasets 
import load_iris from sklearn.preprocessing 
import MinMaxScaler from minisom 
import MiniSom 
# Load Iris dataset 
iris = load_iris() 
X = iris.data 
y = iris.target 
target_names = iris.target_names 
# Scale the data 
scaler = MinMaxScaler() 
X_scaled = scaler.fit_transform(X) 
# Create SOM 
som = MiniSom( 
    x=7,              # number of rows in SOM grid 
    y=7,                # number of columns in SOM grid
    input_len=4,      # number of features in Iris 
    sigma=1.0,)        # neighborhood radius learning_rate=0.5  
# Initialize weights 
som.random_weights_init(X_scaled)
# Train SOM 
som.train_random(    
    X_scaled,    
    1000             # number of iterations 
    ) 
# Plot SOM map 
plt.figure(figsize=(8,8)) for i, x in enumerate(X_scaled):
    winner = som.winner(x)    
plt.text(        
    winner[0],        
    winner[1],        
    str(y[i]),        
    fontsize=12    
    ) 
    plt.xlim([0,7]) 
    plt.ylim([0,7]) 
    plt.grid() 
    plt.title("Self Organizing Map - Iris Dataset") 
    plt.xlabel("SOM X coordinate") 
    plt.ylabel("SOM Y coordinate") 
plt.show()
```

```python
import numpy as np 
import matplotlib.pyplot as plt 
from sklearn.datasets import load_iris 
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis 
# Load the Iris dataset 
iris = load_iris()
#Features 
X = iris.data  
y = iris.target  
# Class labels 
print(X) 
print(y) 
target_names = iris.target_names 
print(target_names) 
# Initialize and fit LDA 
lda = LinearDiscriminantAnalysis(n_components=2) 
X_lda = lda.fit_transform(X, y) 
# Plotting the LDA results 
plt.figure(figsize=(8, 6)) 
colors = ['purple', 'darkgreen', 'yellow'] 
for class_index, class_name in enumerate(target_names):
        plt.scatter(        
        X_lda[y == class_index, 0],  # LD1        
        X_lda[y == class_index, 1],  # LD2        
        label=class_name,        
        alpha=0.7,        
        edgecolors='k',        
        c=colors[class_index]    ) 
        plt.xlabel('LD1') 
        plt.ylabel('LD2') 
        plt.title('LDA of Iris Dataset') 
        plt.legend() 
        plt.grid()
        plt.show()
```

```python
import numpy as np 
import matplotlib.pyplot as plt 
from sklearn.decomposition import PCA
# Create a simple dataset 
X = np.array([[1, 2, 3], [400, 5, 6], [7, 8000, 9], [10, 11, 12111]]) 
# Initialize and fit PCA 
pca = PCA(n_components=2) 
X_pca = pca.fit_transform(X) 
# Print the original dataset 
print("Original Dataset:\n", X) 
# Print the transformed dataset 
print("\nTransformed Dataset:\n", X_pca)
# Print the explained variance ratio 
print("\nExplained Variance Ratio:", pca.explained_variance_ratio_) 
# Visualize the transformed dataset 
plt.scatter(X_pca[:, 0], X_pca[:, 1]) 
plt.title("PCA on Simple Dataset") 
plt.xlabel("PC1") 
plt.ylabel("PC2") 
plt.show()
```
